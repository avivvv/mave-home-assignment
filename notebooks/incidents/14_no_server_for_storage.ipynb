Checked the web:
#### This error indicates that an HPC compute node or client is attempting to access a specific storage domain or volume ("storage X"), but none of the designated storage/metadata servers (MDS or OSS) responsible for that file system domain are currently reachable or active.
#### This typically happens during an outage, network partitioning, or an incomplete storage mount process on a parallel/clustered file system
Common Causes & How to Fix Them
1. Network Connectivity Issues
If nodes cannot communicate over the private cluster network, they cannot reach the configuration servers.
• The Fix: Check the network status between the affected node and the GPFS cluster managers.
2. GPFS Daemon is Down.
The GPFS daemon (mmfsd) might have crashed or failed to start on the configuration server nodes or the local node.
3. Stale Filesystem Mounts or Stale SSL Certificates.
Sometimes, a sudden reboot or network blip leaves stale mount points, or communication is blocked due to expired/mismatched cluster security certificates.
4. Cluster Configuration Desynchronization
The node might have an outdated view of who the primary and secondary cluster servers are.
